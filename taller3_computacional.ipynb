## Ejercicio 3: Recuperacion de mensaje con diagonalizacion

Sea $K$ una matriz llave con valores propios

$$
\lambda_1=1, \quad \lambda_2=3,
$$

y con vector propio asociado a $\lambda_2$:

$$
v_2=\begin{bmatrix}1\\0\end{bmatrix}.
$$

Ademas, la combinacion de letras `BA` permanece igual al codificarse con $K$ (su vector no cambia).

Si el mensaje encriptado es

$$
\begin{bmatrix}
21&9&-37&13&-24&18&15&0&-18&15&57&0&-33&21&-1&4&-69&18&3&0&44&1&18&9&57&0
\end{bmatrix},
$$

entonces, escrito como matriz por bloques $2\times1$, queda

$$
\begin{bmatrix}
21 & -37 & -24 & 15 & -18 & 57 & -33 & -1 & -69 & 3 & 44 & 18 & 57 \\
9 & 13 & 18 & 0 & 15 & 0 & 21 & 4 & 18 & 0 & 1 & 9 & 0
\end{bmatrix}.
$$

a. Escriba un plan para recuperar el mensaje original.
b. Encuentre el mensaje original.
