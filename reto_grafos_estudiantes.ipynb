## Paso 0. Preparar el cuaderno
Ejecuta la celda. Carga NumPy y una función para dibujar la red. **No necesitas entender su código.**

```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=3, suppress=True)   # muestra 3 decimales

# ---- Datos de la red y función para dibujarla (no hace falta modificar nada) ----
aristas = [(1, 2), (1, 3), (1, 4), (2, 3), (2, 4), (3, 4), (4, 5), (5, 6), (5, 7), (5, 8), (5, 9)]
pos = np.array([[0.5, 4.0], [1.5, 4.0], [0.5, 3.1], [1.5, 3.1], [1.5, 2.4],
                [0.7, 2.25], [2.3, 2.25], [2.3, 3.0], [2.7, 2.5]])

def dibujar(tamanos=None, colores=None, titulo=""):
    fig, ax = plt.subplots(figsize=(4.5, 4.5))
    for i, j in aristas:
        ax.plot(*pos[[i - 1, j - 1]].T, color="0.6", lw=1.5, zorder=1)
    t = np.full(9, 500.0) if tamanos is None else 150 + 1200 * np.abs(tamanos) / np.abs(tamanos).max()
    ax.scatter(*pos.T, s=t, c=colores or ["white"] * 9, edgecolors="black", linewidths=1.5, zorder=2)
    for k, (x, y) in enumerate(pos):
        ax.text(x, y, str(k + 1), ha="center", va="center", fontweight="bold", zorder=3)
    ax.set_xlim(0.1, 3.1); ax.set_ylim(1.85, 4.4); ax.set_aspect("equal"); ax.axis("off"); ax.set_title(titulo)
    plt.show()

dibujar(titulo="La red del reto")
```
