## 0.3 · Operar con vectores: elemento a elemento

Los operadores `+`, `-` y `*` de NumPy trabajan **entrada por entrada**. Esto es exactamente la
suma de vectores y el producto por escalar que vimos en clase:

$$\vec u+\vec v=(u_1+v_1,\;\dots,\;u_n+v_n), \qquad
  c\,\vec u=(c\,u_1,\;\dots,\;c\,u_n).$$

````python
bruno = M[1]

print("Ana           :", ana)
print("Bruno         :", bruno)
print("Ana + Bruno   :", ana + bruno)      # suma de vectores
print("2 * Ana       :", 2 * ana)          # producto por escalar
print("Ana - Bruno   :", ana - bruno)
print("Ana * Bruno   :", ana * bruno)      # OJO: esto NO es el producto punto
````

> ⚠️ **El error más común del taller.** `ana * bruno` multiplica entrada por entrada y devuelve
> un **vector**. El producto punto suma esos productos y devuelve **un número**. Son cosas
> distintas:
>
> $$\texttt{ana * bruno} = (20,\,2,\,2,\,20,\,0,\,20) \qquad\text{pero}\qquad
>   \vec a\cdot\vec b = 20+2+2+20+0+20 = 64.$$

## 0.4 · Las tres operaciones del taller

### Producto punto

$$\vec u\cdot\vec v=u_1v_1+u_2v_2+\cdots+u_nv_n$$

En NumPy hay dos formas, y **dan exactamente lo mismo**:

```python
u @ v              # la forma corta (recomendada)
np.sum(u * v)      # la definición, escrita tal cual
```

### Norma

$$\|\vec u\|=\sqrt{\vec u\cdot\vec u}=\sqrt{u_1^2+\cdots+u_n^2}$$

```python
np.linalg.norm(u)          # la forma corta
np.sqrt(np.sum(u**2))      # la definición
```

### Ángulo

$$\cos\theta=\frac{\vec u\cdot\vec v}{\|\vec u\|\,\|\vec v\|}
\qquad\Longrightarrow\qquad
\theta=\arccos\!\left(\frac{\vec u\cdot\vec v}{\|\vec u\|\,\|\vec v\|}\right)$$

```python
cos = (u @ v) / (np.linalg.norm(u) * np.linalg.norm(v))
ang = np.degrees(np.arccos(np.clip(cos, -1, 1)))
```

`np.clip(cos, -1, 1)` recorta el valor al intervalo $[-1,1]$. Hace falta porque el redondeo del
computador puede producir `1.0000000000000002`, y `arccos` de eso no existe.

````python
u = np.array([4, 1])
v = np.array([2, 5])

pp   = u @ v
nu   = np.linalg.norm(u)
nv   = np.linalg.norm(v)
cos  = pp / (nu * nv)
ang  = np.degrees(np.arccos(np.clip(cos, -1, 1)))

print(f"u . v      = {pp}")
print(f"||u||      = {nu:.4f}     (exacto: raíz de 17)")
print(f"||v||      = {nv:.4f}     (exacto: raíz de 29)")
print(f"cos(theta) = {cos:.4f}")
print(f"theta      = {ang:.2f} grados")
print()
print("las dos formas coinciden:",
      np.isclose(u @ v, np.sum(u * v)),
      np.isclose(np.linalg.norm(u), np.sqrt(np.sum(u**2))))
````

### Y así se ve

En $\mathbb{R}^2$ podemos **dibujar** lo que acabamos de calcular. En $\mathbb{R}^6$ ya no, pero
las fórmulas son las mismas: esa es justamente la ventaja de tener una definición algebraica.

````python
fig, ax = plt.subplots(figsize=(5.4, 4.6))
ax.set_axisbelow(True); ax.grid(True, linewidth=0.8)

ax.annotate('', xy=(4, 1), xytext=(0, 0),
            arrowprops=dict(arrowstyle='-|>', color=AZUL, linewidth=2.2))
ax.annotate('', xy=(2, 5), xytext=(0, 0),
            arrowprops=dict(arrowstyle='-|>', color=NARANJA, linewidth=2.2))
ax.text(4.15, 1.0, r'$\vec{u}=(4,1)$', color=AZUL, fontsize=11, fontweight='bold', va='center')
ax.text(2.0, 5.25, r'$\vec{v}=(2,5)$', color=NARANJA, fontsize=11, fontweight='bold', ha='center')

# el arco del ángulo
t = np.linspace(np.arctan2(1, 4), np.arctan2(5, 2), 60)
ax.plot(1.25 * np.cos(t), 1.25 * np.sin(t), color=TENUE, linewidth=1.4)
ax.text(1.42 * np.cos(t.mean()), 1.42 * np.sin(t.mean()),
        f'{ang:.1f}°', color=SUAVE, fontsize=10, ha='center', va='center')

ax.plot(0, 0, 'o', color=TENUE, markersize=5)
ax.set_xlim(-0.6, 5.6); ax.set_ylim(-0.6, 6.0)
ax.set_aspect('equal')
ax.set_xlabel('coordenada 1'); ax.set_ylabel('coordenada 2')
ax.set_title('El ángulo entre dos vectores')
fig.text(0.5, -0.02,
         f'$\\vec{{u}}\\cdot\\vec{{v}}={pp}$  ·  '
         f'$\\|\\vec{{u}}\\|={nu:.3f}$  ·  $\\|\\vec{{v}}\\|={nv:.3f}$  ·  '
         f'$\\cos\\theta={cos:.4f}$',
         ha='center', fontsize=10, color=SUAVE)
plt.tight_layout(); plt.show()
````

## 0.6 · Por qué no hace falta ningún ciclo

Tres herramientas reemplazan todos los `for` de este taller.

**1. El argumento `axis`.** Casi toda función de NumPy acepta `axis`, que significa *«a lo largo
de qué dirección»*: `axis=0` recorre **filas** (resultado por columna, es decir por género) y
`axis=1` recorre **columnas** (resultado por fila, es decir por persona).

**2. El *broadcasting*.** Si una matriz y un vector tienen formas compatibles, NumPy estira el
vector automáticamente: `M - np.array([1,1,1,1,1,1])` le resta $1$ a **todas** las casillas de
golpe. Y si en vez de una fila le da una **columna** de forma `(8,1)`, la estiramos a lo ancho:
entonces la operación se aplica *fila por fila*, cada una con su propio número. Ese truco —
convertir un vector de 8 números en una columna con `[:, None]` — va a aparecer en el Punto 5.

**3. El producto de matrices `@`.** `M @ v` hace, de una sola vez, el producto punto de cada fila
de `M` contra `v`. Y `M @ M.T` da **todos** los productos punto entre todas las parejas.

````python
suma_por_persona = M.sum(axis=1)     # un total por PERSONA (axis=1 recorre los géneros)
tope_por_genero  = M.max(axis=0)     # el máximo por GÉNERO  (axis=0 recorre las personas)

print("puntos que reparte cada persona :", suma_por_persona)
print("calificación más alta por género:", tope_por_genero)
print()
print("M @ ana  (producto punto de cada persona contra Ana):", M @ ana)
print("forma de M @ M.T (todas las parejas de una vez)     :", (M @ M.T).shape)
````

> **Equivalencias que conviene tener a mano**
>
> | Con ciclo (prohibido) | Con NumPy (lo que se espera) |
> |---|---|
> | recorrer y sumar productos | `u @ v` |
> | recorrer y acumular cuadrados | `np.linalg.norm(u)` |
> | una norma por cada fila | `np.linalg.norm(M, axis=1)` |
> | comparar `v` contra todas las filas | `M @ v` |
> | todas las parejas | `M @ M.T` |
> | buscar el máximo recorriendo | `np.argmax(x)` |
> | ordenar recorriendo | `np.argsort(x)` |

# Para cerrar

En una hora usted usó, sobre datos reales y sin escribir un solo ciclo:

| Concepto | Fórmula | NumPy | Qué significó aquí |
|---|---|---|---|
| Suma y escalar | $\vec u+\vec v$, $c\vec u$ | `u + v`, `c * u` | combinar y reescalar perfiles |
| Producto punto | $\vec u\cdot\vec v=\sum u_iv_i$ | `u @ v` | afinidad **sin** normalizar |
| Norma | $\|\vec u\|=\sqrt{\vec u\cdot\vec u}$ | `np.linalg.norm(u)` | qué tan alto califica alguien |
| Ángulo | $\cos\theta=\frac{\vec u\cdot\vec v}{\|\vec u\|\|\vec v\|}$ | `np.arccos(...)` | afinidad **pura** de gustos |
| Ortogonalidad | $\vec u\cdot\vec v=0$ | `u @ v == 0` | gustos sin nada en común |
| Distancia | $\|\vec u-\vec v\|$ | `np.linalg.norm(u-v)` | qué tan parecidas son las notas |

Y la idea que vale más que todas: **el producto punto mezcla dirección con magnitud; dividir por
las normas deja solo la dirección.** Esa división es, literalmente, la que hace que un recomendador
no le sugiera siempre a la persona que más califica.
